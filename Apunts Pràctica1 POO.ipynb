## 1. Herència i Jerarquia de Classes
Permet reutilitzar codi entre classes relacionades.

```python
public class Allotjament {
    protected String nom;
    protected String id;
    public Allotjament(String nom, String id) {
        this.nom = nom;
        this.id = id;
    }
}

public class casa extends Allotjament {
    protected int numHabitacions;
    protected int capacitat;
    public Casa(String nom, String id, int numHabitacions, int capacitat) {
        super(nom, id);
        this.numHabitacions = numHabitacions;
        this.capacitat = capacitat;
    }
}
```

Casa hereta Allotjament, reutilitza atributs i mètodes

## 2. Encapsulació i Modificadors d'Accés
Controla la visibilitat d'atributs/mètodes

```python
public class Glamping {
    private boolean casaMascota;
    
    public boolean getCasaMascota() {
        return casaMascota;
    }
    
    public void setCasaMascota(boolean casaMascota) {
        this.casaMascota = casaMascota;
    }
}
```

**private -->** només accessible dins la classe. <br>
**public -->** accessible des de fora.

## 3. Sobreescriptura de Mètodes (@Override)
Canvia el comportament d'un mètode heretat

```python
public class Bungalow extends Casa {  
    @Override  
    public String getNom() {  
        return "Bungalow: " + super.nom;  
    }  
}
```

**@Override** indica que modifica un mètode existent.

## 4. Polimorfisme i Tipus de Refèrencia

Permet tractar onjectes de diferents subclasses com la seva superclasse.

```python
Allotjament bung = new Bungalow("Bungalow Nord", "B001", 2, 4, 2, 4, 1, true, true, true);
if (bung instanceof Bungalow) {
    System.out.println("És un Bungalow");
}
```

**instanceof** verifica el tipus en temps d'execució.
